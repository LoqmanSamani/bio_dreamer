# 04 — Active Learning & Experiment Design

**Purpose:** Demonstrate the active learning loop where the model proposes experiments, receives (simulated) wet-lab feedback, and updates.

## Sections

### 1. Setup
- Load trained world model + policy
- Initialise ActiveLearner with acquisition function (MaxUncertainty, UCB, or BatchDiversity)

### 2. Cycle 1: Dream → Propose
- Dream candidate variants
- Rank by expected information gain
- Select batch of K experiments to propose
- Visualise uncertainty landscape before feedback

### 3. Simulate Wet-Lab Feedback
- Use oracle (DMS lookup / ESMFold / random noise) as ground-truth
- Generate pseudo-experimental fitness values

### 4. Update Model
- Fine-tune world model on new data points
- Show uncertainty reduction after incorporation
- Compare model predictions before/after update

### 5. Iterate
- Run 3-5 active learning cycles
- Plot: Spearman ρ vs. cycle, uncertainty vs. cycle, cumulative data points
- Demonstrate convergence toward optimal design

### 6. Discussion
- Compare acquisition functions
- Analyse which mutations were most informative
