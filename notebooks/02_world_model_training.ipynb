# 02 — World Model Training

**Purpose:** Step-by-step world model training with interactive monitoring.

## Sections

### 1. Configuration
- Load and inspect YAML config for chosen module
- Override hyperparameters inline for experimentation

### 2. Data Loading
- Initialise dataset and DataLoader
- Inspect a training batch: shapes, dtypes, value ranges

### 3. Model Instantiation
- Create encoder, dynamics, decoder, reward head
- Compose into WorldModel
- Print parameter counts per component

### 4. Training Loop
- Run WorldModelTrainer.fit() with inline loss plotting
- Monitor: reconstruction loss, dynamics KL, reward MSE
- Visualise latent space (t-SNE/UMAP of z_t) every N epochs

### 5. Evaluation
- Dream rollouts on held-out data
- Compare dreamed vs. actual next-states
- Compute Spearman ρ on fitness predictions

### 6. Checkpoint
- Save trained model
- Demonstrate loading and verifying output
