# 05 — Hugging Face Hub Integration

**Purpose:** Demonstrate uploading trained models to Hugging Face Hub and downloading community/custom models.

## Sections

### 1. Browse Available Models
- Use ModelRegistry to list pre-registered community models
- Display model cards for ESM-2, ESMFold, Geneformer

### 2. Download a Community Model
- Download ESM-2 (or a small variant) from HF Hub
- Show caching behaviour (second call is instant)
- Load model weights into BioDreamer encoder

### 3. Upload Custom Model
- Take trained world model from notebook 02
- Generate model card with auto-filled metrics
- Export to safetensors format
- Push to HF Hub (private repo)
- Verify upload by downloading and comparing outputs

### 4. Version Management
- Upload v2 of the same model
- Download specific revision (v1 vs. v2)
- Compare model outputs across versions

### 5. Model Card Customisation
- Edit auto-generated model card
- Add training curves, evaluation tables
- Preview rendered markdown
