# 01 — Data Exploration & Preprocessing

**Purpose:** Interactive exploration of datasets used across all three BioDreamer modules.

## Sections

### 1. ProteinDreamer Data
- Load ProteinGym substitution benchmark CSVs
- Visualise fitness distributions, mutation depth histograms
- Inspect Tsuboyama mega-scale DMS data
- Compute dataset statistics (sequence lengths, mutation counts)

### 2. MolDreamer Data
- Load sample MD trajectory (ATLAS or D.E. Shaw)
- Visualise RMSD, energy over frames
- Plot 3-D structure snapshots with py3Dmol

### 3. CellDreamer Data
- Load Norman 2019 Perturb-seq h5ad with scanpy
- QC plots: gene counts, mitochondrial fraction
- UMAP coloured by cell type and perturbation
- Gene expression distributions for key markers

### 4. Preprocessing Pipelines
- Demonstrate preprocessing functions from each module's data/ subpackage
- Verify output shapes match model input requirements
